Testing score(), recommend() and explain() from src/portfolio_projekt/recommend.py.

```python
%load_ext autoreload
%autoreload 2

import pandas as pd

from portfolio_projekt.recommend import explain, load_bundle, recommend, score
```

```python
examples = [
    "Thanks for the article, very informative.",
    "I strongly disagree with this policy.",
    "This is a stupid idea.",
    "You are an idiot and should shut up.",
    "If I see you again, you will regret it.",
    "I am a proud gay man.",
    "Muslims in my neighborhood organized a food drive.",
]

thresholds = load_bundle()["thresholds"]
rows = []
for text in examples:
    scores = score(text)
    result = recommend(scores, thresholds)
    rows.append({
        "text": text,
        "toxicity": round(scores["toxicity"], 2),
        "level": result["level"],
        "action": result["action"],
        "flagged": ", ".join(result["flagged"]),
    })

pd.set_option("display.max_colwidth", None)
pd.DataFrame(rows)
```
